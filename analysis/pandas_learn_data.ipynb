```python
import pandas
df = pandas.read_csv("tmdb.csv")
df.isnull() # 查看数据中是否有缺失值
df.dropna() # 删除数据中缺失值所在的行
df.ffill() # 用上一行的数据填充缺失值
df.bfill() # 用下一行的数据填充缺失值
df.fillna("--") # 用指定值填充缺失值

df.duplicated() # 查看数据中是否有重复数据
df.duplicated(subset=["年份", "评分"]) # 查看数据中指定列是否有重复数据
df.drop_duplicates(subset=["年份", "评分"]) # 删除数据中指定列的重复数据

df[df["时长"] >= 300] #查看异常值
df.drop(df[df["时长"] >= 300].index) # 删除异常值
df["时长"] = df["时长"].abs() # 处理异常值

df["时长"].astype("str").str.replace("$","分钟",regex=True)
```

```python
df.sort_values("年份",ascending= False)
df.sort_values(["年份","评分"],ascending=[False,True])
```

```python
df.groupby("年份")["电影名"].count()
df.groupby("年份")["评分"].sum()
df.groupby("年份")["评分"].min()
df.groupby("年份")["评分"].max()
df.groupby("年份")["评分"].mean()
df.groupby("年份")["评分"].aggregate(["count","sum","min","max","mean"])
df.groupby("年份").aggregate({"评分":["count","sum","min","max","mean"],
                              "时长":["count","sum","min","max","mean"]})
```
