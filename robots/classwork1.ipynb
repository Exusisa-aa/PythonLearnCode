```python
import csv
import requests
import time
import random
import os

if __name__ == '__main__':
    # 定义表头
    header = ['昵称', '年龄', '身高', '地区', '学历', '收入', '婚姻状况', '自我介绍', '头像链接']
    # 图片保存的文件夹
    img_dir = "csv_data/classwork1pic"
    # 如果文件夹不存在则创建
    if not os.path.exists(img_dir):
        os.makedirs(img_dir)
    # 创建csv文件
    with open("csv_data/classwork1.csv", "w", encoding="UTF-8", newline="") as f:
        # 创建操作的csv文件对象，并指定文件对象和表头列表
        writer = csv.DictWriter(f, fieldnames=header)
        # 写入表头
        writer.writeheader()

        # 当前年份，用于根据出生年份计算年龄
        now_year = time.localtime().tm_year
        # 页码，从第一页开始
        page = 1

        while page <= 30:
            # 缘分推荐接口地址（页面中的“缘分推荐”内容由该接口动态加载）
            url = "https://www.7799520.com/api/recommend/pc/luck/list"
            # 发送带参数请求，拿到json数据
            res = requests.request("GET", url, params={"userid": "", "token": "", "page": page})
            user_list = res.json().get("data", {}).get("list", [])

            # 列表为空说明数据已经爬取完毕，跳出循环
            if not user_list:
                print("数据爬取完毕，共爬取 %d 页" % (page - 1))
                break

            # 获取每个用户的推荐数据
            for user in user_list:
                # 用户id
                user_id = user.get("userid", "")

                # 昵称
                user_name = user.get("username", "")
                print("已获取昵称：" + user_name)

                # 年龄（用当前年份减去出生年份得到）
                birthday_year = user.get("birthdayyear", "")
                if birthday_year:
                    user_age = str(now_year - int(birthday_year))
                else:
                    user_age = ""
                print("已获取年龄：" + user_age)

                # 身高
                user_height = user.get("height", "")
                if user_height:
                    user_height = user_height + "CM"
                print("已获取身高：" + user_height)

                # 地区（省 + 市）
                user_city = (user.get("province", "") + user.get("city", "")).strip()
                print("已获取地区：" + user_city)

                # 学历
                user_education = user.get("education", "")
                print("已获取学历：" + user_education)

                # 收入
                user_salary = user.get("salary", "")
                print("已获取收入：" + user_salary)

                # 婚姻状况
                user_marry = user.get("marry", "")
                print("已获取婚姻状况：" + user_marry)

                # 自我介绍
                user_monolog = user.get("monolog", "")
                print("已获取自我介绍：" + user_monolog)

                # 头像链接
                user_avatar = user.get("avatar", "")
                print("已获取头像链接：" + user_avatar)

                # 下载头像图片
                if user_avatar:
                    # 发送请求下载图片
                    img_res = requests.request("GET", user_avatar)
                    # 根据响应头判断图片后缀
                    content_type = img_res.headers.get("Content-Type", "")
                    if "png" in content_type:
                        suffix = ".png"
                    elif "gif" in content_type:
                        suffix = ".gif"
                    else:
                        suffix = ".jpg"
                    # 保存图片
                    with open(img_dir + "/" + user_id + suffix, "wb") as img_f:
                        img_f.write(img_res.content)
                    print("已下载头像：" + user_id + suffix)

                body = [user_name, user_age, user_height, user_city, user_education,
                        user_salary, user_marry, user_monolog, user_avatar]

                # 写入文件
                writer.writerow(dict(zip(header, body)))
                print("已写入文件：" + user_name)

            page += 1
            time.sleep(random.uniform(1, 3))
```
