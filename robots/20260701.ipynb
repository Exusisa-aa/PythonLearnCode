```python
import os
import requests
import jsonpath
from urllib.request import urlretrieve
if not os.path.exists('./fig'):
    os.mkdir('./fig')
url = 'https://www.7799520.com/api/recommend/pc/luck/list?token=&page=4'
req = requests.get(url).json()
#print(req)
avatars = jsonpath.jsonpath(req,'$..avatar')
names = jsonpath.jsonpath(req,'$..username')
#print(avatars,names)
for avatar,name in zip(avatars,names):
    print(avatar,name)
    urlretrieve(avatar,'./fig' + '/' + name + '.png')
```

```python
from aip import AipFace
import base64

def get_face(file_path):
    APP_ID = ''
    API_KEY = ''
    SECRET_KEY = ''

    client = AipFace(APP_ID, API_KEY, SECRET_KEY)

    with open(file_path,'rb') as fp:
        data = base64.b64encode(fp.read())

    image = data.decode()
    
    imageType = "BASE64"
    options = {}
    options["face_field"] = "beauty"
    rs = client.detect(image, imageType, options)
    print(rs)
    return rs['result']['face_list'][0]['beauty']
#print(get_face(r'C:\Users\Lenovo\fig\下载.jpg'))
```

```python
import time
path = r'C:\Users\User\fig'
images = os.listdir(path)
#print(images)
yz = []
yz_dict = {}
for image in images:
    score = get_face(path + '/' + image)
    time.sleep(1)
    name = image[0:-4]
    yz.append(score)
    yz_dict[score] = name
yz.sort(reverse=True)
```

```python
for a,b in enumerate(yz):
    print('小姐姐的名字是：{}=====她是第{}名=====她的颜值分是{}'.format(yz_dict[b],a+1,b))
```
